## Handling Missing Values

The `Income` column contained 24 missing values. Since the income distribution was right-skewed with extreme outliers, the median was chosen to impute the missing values. The median is less affected by outliers than the mean, making it a more robust measure of central tendency.


```python

df["Income"] = df["Income"].fillna(df["Income"].median())

missing = df.isnull().sum()

missing[missing > 0]
```
